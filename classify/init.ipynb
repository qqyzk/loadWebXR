```python
# Import required libraries
from bs4 import BeautifulSoup
import re

# Set file path - modify to your actual HTML file path
file_path = "../realCase/EllieTale-WebXR-main/html/Beach.html"  # Replace with your HTML file path

# Read HTML file content
with open(file_path, 'r', encoding='utf-8') as file:
    html_content = file.read()

# Create BeautifulSoup parser
soup = BeautifulSoup(html_content, 'html.parser')

# Counter
total_assets = 0
updated_models = 0
already_high_priority = 0
audio_resources = 0

# Process all <a-asset-item> elements
for item in soup.find_all('a-asset-item'):
    total_assets += 1
    src = item.get('src', '')
    
    # Check if it's a model file (.glb or .gltf)
    if src and re.search(r'\.(glb|gltf)$', src, re.IGNORECASE):
        # Check if priority attribute already exists
        if item.has_attr('priority'):
            # If already highest priority
            if item['priority'] == '1':
                already_high_priority += 1
            # If other priority, update to 1
            else:
                item['priority'] = '1'
                updated_models += 1
        # If no priority attribute, add priority="1"
        else:
            item['priority'] = '1'
            updated_models += 1
            
    # Count audio files
    elif src and re.search(r'\.(wav|mp3|ogg)$', src, re.IGNORECASE):
        audio_resources += 1

# Generate modified HTML content
modified_html = soup.prettify()

# Save to new file - modify to your desired output path
output_path = "../realCase/EllieTale-WebXR-main/html/priority_updated_scene.html"
with open(output_path, 'w', encoding='utf-8') as file:
    file.write(modified_html)

# Display processing results
print("="*50)
print(f"Resource Statistics:")
print(f"- Total assets found: {total_assets}")
print(f"- Model resources: {updated_models + already_high_priority}")
print(f"- Audio resources: {audio_resources}")
print(f"- Other resource types: {total_assets - updated_models - already_high_priority - audio_resources}")
print("\nPriority Processing Results:")
print(f"- Updated priority for {updated_models} model resources to highest (priority=1)")
print(f"- {already_high_priority} model resources already had highest priority")
print(f"- All audio resources maintain their original priority settings")
print(f"\nOutput file saved to: {output_path}")
print("="*50)
```
