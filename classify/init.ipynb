```python
# 导入所需库
from bs4 import BeautifulSoup
import re

# 设置文件路径 - 修改为您的实际HTML文件路径
file_path = "../realCase/EllieTale-WebXR-main/html/Beach.html"  # 替换为您的HTML文件路径

# 读取HTML文件内容
with open(file_path, 'r', encoding='utf-8') as file:
    html_content = file.read()

# 创建BeautifulSoup解析器
soup = BeautifulSoup(html_content, 'html.parser')

# 计数器
total_assets = 0
updated_models = 0
already_high_priority = 0
audio_resources = 0

# 处理所有<a-asset-item>元素
for item in soup.find_all('a-asset-item'):
    total_assets += 1
    src = item.get('src', '')
    
    # 检查是否为模型文件（.glb或.gltf）
    if src and re.search(r'\.(glb|gltf)$', src, re.IGNORECASE):
        # 检查是否已有priority属性
        if item.has_attr('priority'):
            # 如果已经是第一优先级
            if item['priority'] == '1':
                already_high_priority += 1
            # 如果是其他优先级，更新为1
            else:
                item['priority'] = '1'
                updated_models += 1
        # 如果没有priority属性，添加priority="1"
        else:
            item['priority'] = '1'
            updated_models += 1
            
    # 统计音频文件
    elif src and re.search(r'\.(wav|mp3|ogg)$', src, re.IGNORECASE):
        audio_resources += 1

# 生成修改后的HTML内容
modified_html = soup.prettify()

# 保存到新文件 - 修改为您想要的输出路径
output_path = "../realCase/EllieTale-WebXR-main/html/priority_updated_scene.html"
with open(output_path, 'w', encoding='utf-8') as file:
    file.write(modified_html)

# 显示处理结果
print("="*50)
print(f"资源统计:")
print(f"- 总共找到 {total_assets} 个资源")
print(f"- 其中模型资源: {updated_models + already_high_priority} 个")
print(f"- 其中音频资源: {audio_resources} 个")
print(f"- 其他类型资源: {total_assets - updated_models - already_high_priority - audio_resources} 个")
print("\n优先级处理结果:")
print(f"- 已更新 {updated_models} 个模型资源的优先级为第一优先级 (priority=1)")
print(f"- {already_high_priority} 个模型资源已经是第一优先级")
print(f"- 所有音频资源保持原有优先级设置不变")
print(f"\n输出文件已保存至: {output_path}")
print("="*50)
```
